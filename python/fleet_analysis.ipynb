```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
vehicles = pd.read_csv(
    "fleet_vehicles.csv"
)

dispatch = pd.read_csv(
    "emergency_dispatch.csv"
)
```

The project uses two operational datasets:

1. Fleet vehicle information
2. Emergency dispatch records

```python
fleet_data = dispatch.merge(
    vehicles,
    on="vehicle_id",
    how="left"
)
```

```python
average_response_time = (
    fleet_data["response_minutes"]
    .mean()
)

print(
    "Average Response Time:",
    round(average_response_time,2),
    "minutes"
)
```

Average response time measures how quickly emergency resources reach incidents.

Lower response times generally indicate better operational efficiency.

```python
vehicle_workload = (

    fleet_data
    .groupby("vehicle_id")
    .agg(
        total_dispatches=("incident_id","count"),
        avg_response_time=("response_minutes","mean")
    )
    .reset_index()

)

vehicle_workload.head()
```

```python
vehicle_workload.sort_values(
    "total_dispatches",
    ascending=False
).head(10)
```

```python
station_analysis = (

    fleet_data
    .groupby("home_station")
    .agg(
        total_incidents=("incident_id","count"),
        avg_response_time=("response_minutes","mean"),
        vehicles=("vehicle_id","nunique")
    )
    .reset_index()

)

station_analysis
```

```python
station_analysis["dispatches_per_vehicle"] = (

    station_analysis["total_incidents"]

    /

    station_analysis["vehicles"]

)

station_analysis
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=vehicle_workload.sort_values(
        "total_dispatches",
        ascending=False
    ).head(10),

    x="vehicle_id",

    y="total_dispatches"
)

plt.xticks(rotation=45)

plt.title(
    "Top 10 Vehicles by Dispatch Workload"
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    fleet_data["response_minutes"],
    bins=20
)

plt.title(
    "Emergency Response Time Distribution"
)

plt.xlabel(
    "Response Minutes"
)

plt.show()
```

# Key Findings

The analysis evaluates emergency fleet utilization using operational KPIs.

Main observations:

- Vehicles can be compared based on dispatch workload.
- Stations can be evaluated using workload per vehicle.
